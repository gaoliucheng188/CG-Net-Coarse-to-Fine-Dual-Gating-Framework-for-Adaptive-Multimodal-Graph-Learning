```python
import pickle
import os
import networkx as nx

path = 'data/tud_mutag.pkl'
if os.path.exists(path):
    with open(path, 'rb') as f:
        data = pickle.load(f)
    print(f"Type of data: {type(data)}")
    if isinstance(data, list):
        print(f"Length of list: {len(data)}")
        print(f"Type of first element: {type(data[0])}")
        if len(data) > 0 and isinstance(data[0], nx.Graph):
             print("First element is a NetworkX Graph")
             print(f"Nodes: {data[0].number_of_nodes()}")
             print(f"Edges: {data[0].number_of_edges()}")
             print(f"Node 0 label: {data[0].nodes[0].get('label')}")
    elif isinstance(data, tuple):
        print(f"Tuple length: {len(data)}")
        print(f"Type of elements: {[type(x) for x in data]}")
else:
    print(f"File {path} not found.")
```
